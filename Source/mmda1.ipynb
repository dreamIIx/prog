```python
from sklearn import datasets

iris = datasets.load_iris()
print(iris['DESCR'])

iris.__dir__()

print('The unique labels (targets) in this data are', str(np.unique(iris.target)))
print("Labels names", list(iris.target_names))

target_idx_to_name = dict(zip(np.unique(iris.target), iris.target_names))
print(target_idx_to_name)

labels = pd.DataFrame(iris.target, columns=['species']).applymap(lambda t: target_idx_to_name[t])
iris_full = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_full = iris_full.join(labels)
print(iris_full)
```

```python
X, y = iris.data, iris.target
print(X)
print(y)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.3, random_state=0)

print('There are {} samples in the training set and {} samples in the test set'.format(
X_train.shape[0], X_test.shape[0]))
```

```python
import itertools

def plot_data(X_train, X_test, y_train, y_test, fit_f, predict_f, score_f, markers, colors):
  fit_f(X_train, y_train)
  y_pred_4d = predict_f(X_test)
  print('The accuracy of the Linear Discriminant classifier on 4D training data is {:.2f} %'.format(100 * score_f(X_train, y_train)))
  print('The accuracy of the Linear Discriminant classifier on 4D test data is {:.2f} %'.format(100 * np.mean(y_test == y_pred_4d)))

  train_acc_2d, test_acc_2d = [], []
  fig, axes = plt.subplots(2, 3, figsize = (18, 12))
  plt.subplots_adjust(wspace = 0.4, hspace = 0.4)

  for i, (feat1, feat2) in enumerate(itertools.combinations(range(4), 2)):
      X_train_subset = X_train[:, [feat1, feat2]]
      X_test_subset = X_test[:, [feat1, feat2]]

      x1_min, x1_max = X_test_subset[:, 0].min() - 1, X_test_subset[:, 0].max() + 1
      x2_min, x2_max = X_test_subset[:, 1].min() - 1, X_test_subset[:, 1].max() + 1
      xx1, xx2 = np.meshgrid(
          np.arange(x1_min, x1_max, step = 0.02),
          np.arange(x2_min, x2_max, step = 0.02)
      )

      fit_f(X_train_subset, y_train)
      y_pred = predict_f(X_test_subset)
      Z = predict_f(np.array([xx1.ravel(), xx2.ravel()]).T)
      Z = Z.reshape(xx1.shape)

      ax = axes[i // 3, i % 3]
      ax.set_xlim(x1_min, x1_max)
      ax.set_ylim(x2_min, x2_max)
      ax.set_xlabel(iris.feature_names[feat1])
      ax.set_ylabel(iris.feature_names[feat2])

      train_acc = round(100 * score_f(X_train_subset, y_train), 2)
      test_acc = round(100 * score_f(X_test_subset, y_test), 2)
      ax.set_title(f"train_acc: {train_acc}, test_acc: {test_acc}")

      train_acc_2d.append(train_acc)
      test_acc_2d.append(test_acc)

      ax.contour(xx1, xx2, Z, alpha = 0.4, colors = colors)
      for idx, (x, y) in enumerate(X_test_subset):
          ax.scatter(
              x = x, y = y,
              alpha = 0.8, c = colors[y_pred[idx] - 1],
              marker = markers[y_pred_4d[idx] - 1]
          )

  print()
  print('Mean accuracy of the Linear Discriminant classifier by all 2D training data is {:.2f} %'.format(np.mean(train_acc_2d)))
  print('Meane accuracy of the Linear Discriminant classifier by all 2D test data is {:.2f} %'.format(np.mean(test_acc_2d)))
  print()

  plt.show()
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

markers = ('s', 'x', 'o')
colors = ('red', 'blue', 'lightgreen')

lda = LinearDiscriminantAnalysis()
lda_2d = LinearDiscriminantAnalysis()

plot_data(
  X_train,
  X_test,
  y_train,
  y_test,
  lambda x, y: lda_2d.fit(x, y),
  lambda x: lda_2d.predict(x),
  lambda x, y: lda_2d.score(x, y),
  markers,
  colors)
```

```python
from numpy.linalg import inv, det
import math
import copy

# iris_groupby = iris_two.groupby('species')
# groups = iris_groupby.groups
# iris_two_clear = iris_full[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']]
# print(iris_two_clear.cov())

def substract_mean(group_point, g_mean):
    # print(group_point)
    for i in range(len(group_point)):
        # print(i)
        # print(group_point.iloc[i])
        group_point[i] = group_point[i] - g_mean.iloc[i]

groups_idx = {1: 'versicolor', 2: 'virginica'}
group_names = {'versicolor': 1, 'virginica': 2}

class QDA(object):
    def __init__(self):
        self.g_mean = pd.Series([])
        self.lens = {}
        self.prob_vector = {}
        self.mean_sets = {}
        self.cov_sets = {}
        self.groups = {'versicolor': pd.Index([]), 'virginica': pd.Index([])}

    def my_fit(self, _x_train, _y_train):
        x_train = copy.deepcopy(_x_train)
        y_train = copy.deepcopy(_y_train)
        self.lens = {}
        self.prob_vector = {}
        self.mean_sets = {}
        self.cov_sets = {}
        # print('xlen ', len(x_train))
        # print('x_train ', x_train)
        # print('ylen ', len(y_train))
        # print('y_train ', y_train)
        iris_two_clear = pd.DataFrame(x_train)
        self.g_mean = iris_two_clear.mean()
        # print(iris_two_clear)
        self.groups = {'versicolor': pd.Index([]), 'virginica': pd.Index([])}
        groupz = {'versicolor': [], 'virginica': []}
        idx_ = 0
        for i in y_train:
            # print('i ', i)
            # print('idx_ ', idx_)
            groupz[groups_idx[i]].append(idx_)
            # print('1 ', groupz[groups_idx[1]])
            # print('2', groupz[groups_idx[2]])
            idx_ += 1
        self.groups[groups_idx[1]] = pd.Index(groupz[groups_idx[1]])
        self.groups[groups_idx[2]] = pd.Index(groupz[groups_idx[2]])

        for k, v in self.groups.items():
            # print(k)
            # print(v)
            # print(iris_two_clear.iloc[groups[k]])
            self.lens[k] = len(v.values)
            # print(self.lens[k])
            self.prob_vector[k] = self.lens[k] / len(x_train)
            # print(self.prob_vector[k])
            # print(iris_two_clear.iloc[groups[k]].mean().values)
            self.mean_sets[k] = iris_two_clear.iloc[v.values].mean().values
            # print(self.mean_sets[k])

            for i in v:
                # print(i)
                substract_mean(x_train[i], self.g_mean)
            # print(iris_two_clear.iloc[groups[k]].cov().values)
            self.cov_sets[k] = iris_two_clear.iloc[v.values].cov().values

        # print(cov_sets)

        for k, v in self.cov_sets.items():
            self.cov_sets[k] = inv(self.cov_sets[k])

        # print(self.cov_sets)

    def difference_vectors(self, x, y):
        result = []
        for i in range(len(x)):
            result.append(x[i] - y[i])
        return result

    def dot(self, x, y):
        return sum(a * b for a, b in zip(x, y))

    def vec_on_mtx_prod(self, ui, cvg):
        r = []
        for i, a in enumerate(ui):
            r.append(self.dot(ui, cvg[i]))
        return r

    def scalar_prod(self, scalar, vector):
        result = []
        for i, a in enumerate(vector):
            result.append(vector[i] * scalar)
        return result

    def my_predict(self, _x_test):
        x_test = copy.deepcopy(_x_test)
        predictions = {}
        y_pred = []
        # vv = pd.DataFrame(x_test)
        for idx in range(len(x_test)):
            # v = vv.iloc[idx]
            v = x_test[idx]
            for k, _ in self.groups.items():
                # print(k)
                # print(v)
                # print(self.cov_sets)
                predictions[k] = self.dot(self.vec_on_mtx_prod(self.scalar_prod(-0.5, self.difference_vectors(v, self.mean_sets[k])), \
                    self.cov_sets[k]), self.difference_vectors(v, self.mean_sets[k])) - \
                    (0.5 * math.log(det(self.cov_sets[k]))) + math.log(self.prob_vector[k])
            # print('? ', max(predictions, key=predictions.get))
            # if max(predictions, key=predictions.get) != group_names[y_test[idx] - 1]:
            #     print('failure!')
            # print(predictions)
            y_pred.append(group_names[max(predictions, key=predictions.get)])
        return np.array(y_pred)
```
